### Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mais de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clientes inativos, ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

Base de dados e arquivos: https://drive.google.com/drive/folders/1uDesZePdkhiraJmiyeZ-w5tfc8XsNYFZ?usp=drive_link

```python
# Passo a passo do projeto
#tabula converte de pdf para tabela
# Passo 01: Abri a base de dados (importar a base de dados)
# pip install pandas openpxl ipykernel nbformat plotly
import pandas as pd #para escrever menos, apelido para qualquer pacote
import plotly.express as px #Adicione essa linha antes de usar px.histogram
#!pip install plotly # Plotly Express (importado como px) é um módulo
# da biblioteca Plotly para Python, feito para
# criar gráficos interativos de forma rápida e com poucas linhas de código.
tabela  = pandas.read_csv("cancelamentos.csv")
display(tabela)

#dataframe pandas - tabela bandas

# Passo 02: Vizualizar a base de dados (identificar se tem as informações)
    # entender as informações disponíveis
    # possíveis problemas/erros na base de dados
    #informações que não te ajudam te atrapalham
    #KAGGLE BASE DE DADOS QUE AS EMPERSAS DISPONIBILIZAM PRA PESSOAS TREINAREM
    # INFORMAÇÕES VAZIAS

    #como tirar a coluna
    #lista de colunas > em colchetes lista_nome = ["Lira" ,"Talia","José"]
    #forma rápida, corrigir caracter 
tabela = tabela.drop(columns="CustomerID")
display(tabela)

```

```python
 
# Passo 03: corrigir os problemas da base de dados ( tratamento da base de dados) (adicionar anonimização)
display(tabela.info())

#nformações vazias
tabela = tabela.dropna() #NaN not a number = vazio = valores vazios
#para excluir valores duplicados drop_duplicates()
 
# Passo 03: corrigir os problemas da base de dados ( tratamento da base de dados) (adicionar anonimização)
display(tabela.info())

# COMO IDENTIFICO? Como analiso ?
#Pode tratar a informação, modificando ela de alguma maneira inteligente ou excluindo ela.
#identificar os problema
```

```python
# Passo 04: Análise inicial (entender quantos clientes cancelaram)

#quantidade - contar quantos caras são 0 e quantos são 1 na coluna cancelou
display(tabela["cancelou"].value_counts())
#min média

#porcentagem
display(tabela["cancelou"].value_counts(normalize=True)) #se eu rodar vai dizer o percentual deles


#display(tabela["cancelou"].value_counts(normalize=True).map("{:.1%}".format)) #código de formatação
```

```python
# Passo 05: análise detalhada para entender a causa do cancelamento, como cada coluna impacta no cancelamento
# primeiro cria o gráfico
grafico = px.histogram(tabela, x="cancelou")
# eixo y quantidade de pessoas, x = pessoas que cancelaram

#exibe o gráfico

grafico.show()
```
